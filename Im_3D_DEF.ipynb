```python
import sys
import numpy as np
from matplotlib import pyplot as plt
import matplotlib.gridspec as gridspec
from scipy.constants import e, m_e, pi, hbar
```

# Compute the Imaginary part of the static Dielectric Function for 3-D case

$$
\Im[ϵ(ω)]=2\left(2 m_r\right)^{\frac{3}{2}}\left|p_{v c}\right|^2 \frac{\sqrt{\omega^2-\epsilon_g^2}}{\omega^2} \quad  for ~\omega>\epsilon_g
$$

```python
#Input Parameters

mm_r =  0.067                 # Reduced mass
mm_e = 1                      # Electron mass
mm_r =  mm_r * mm_e
p_vc = 0.9                    # Matrix Element
h2ev = 27.21138386            # Hartree to eV
au2fs = 0.02418884254         # a.u to femtosecond

ww_min = 0.00                 # Frequency Min
ww_max = 2                    # Frequency Max
ww_grid = 100000
ee_g = 4.45/h2ev              # Bandgap in a.u
c = 137.0                     # Speed of light in a.u
f  = 4.45/h2ev  #f Frequency
```

```python
freq = np.linspace(ww_min, ww_max, ww_grid)       # Frequency grid
```

```python
eps = np.zeros(len(freq))
```

```python
for idx, omega in enumerate(freq):
    if omega > ee_g:
        eps[idx] = (
            2.0
            * ((2 * mm_r) ** 3.0 / 2.0)
            * (np.abs(p_vc) ** 2)
            * ((np.sqrt(omega ** 2.0 - ee_g**2.0)) / omega ** 2.0)
        )
```

#Plot of Epsilon

```python
plt.plot(freq*h2ev, eps, 'r', linewidth =2)
plt.xlabel(r'$\omega (eV)$', size = 15)
plt.ylabel(r'$\Im(\epsilon(\omega)) (a.u)$', size = 15)
plt.title(r'Plot of $\Im(\epsilon(\omega))$', size = 15)
plt.xlim(0,10)
```
